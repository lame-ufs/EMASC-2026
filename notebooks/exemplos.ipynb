## 1. Variáveis

O que são? Variáveis são nomes associados a valores armazenados durante a execução do programa. Em Python, não é necessário declarar previamente o tipo da variável.

```python
nome_da_variavel = valor
```

````python
nome = "Carlos"
idade = 20
altura = 1.75

print(nome)
print(idade)
print(altura)
````

Neste exemplo, nome guarda um texto, idade guarda um inteiro e altura guarda um número de ponto flutuante. O sinal `=` atribui um valor à variável.

### Regras de nomes

- O nome começa por uma letra ou por `_`. Depois, pode conter letras, números e `_`.
- Não use espaços, hífens ou números no início do nome.
- Python diferencia maiúsculas de minúsculas: `idade` e `Idade` são nomes distintos.
- Palavras reservadas da linguagem, como `if`, `else`, `for`, `while`, `def`, `return`, `True`, `False` e `None`, não podem ser usadas como nomes de variáveis.

**Válidos:** `nota_final`, `aluno2`, `_total`.

**Inválidos:** `2x`, `nome completo`, `a-b`, `if`.

## 2. Tipos de variáveis

O que são? O tipo indica a categoria do dado armazenado. Os tipos básicos mais comuns são:

| Tipo | Descrição | Exemplo |
| --- | --- | --- |
| `int` | Número inteiro | `idade = 18` |
| `float` | Número de ponto flutuante | `altura = 1.75` |
| `str` | Texto (string) | `nome = "Ana"` |
| `bool` | Valor lógico: verdadeiro ou falso | `aprovado = True` |

Estrutura: atribua um valor e utilize `type()` para descobrir seu tipo.

```python
variavel = valor
type(variavel)
```

````python
idade = 18
altura = 1.70
nome = "João"
aprovado = True

print(type(idade))
print(type(altura))
print(type(nome))
print(type(aprovado))
````

A função `type()` retorna o tipo do objeto. Os resultados serão, respectivamente, `int`, `float`, `str` e `bool`.

**Observação:** `eval()` avalia uma expressão Python escrita em texto, como `eval("2 + 3")`, que retorna 5. Não use com entradas do usuário, pois pode executar código. Para conversões numéricas, prefira `int()` ou `float()`.

## 4. Operadores

O que são? Operadores realizam cálculos, comparações e combinações lógicas.

### 4.1 Operadores aritméticos

| Operador | Operação |
| --- | --- |
| `+` | Adição |
| `-` | Subtração |
| `*` | Multiplicação |
| `/` | Divisão |
| `//` | Divisão inteira (quociente arredondado para baixo) |
| `%` | Resto da divisão |
| `**` | Potenciação |

```python
resultado = valor1 operador valor2
```

````python
numero1 = 10
numero2 = 5

print(numero1 + numero2) # 15
print(numero1 - numero2) # 5
print(numero1 * numero2) # 50
print(numero1 / numero2) # 2.0
print(numero1 // numero2) # 2
print(numero1 % numero2) # 0
print(numero1 ** numero2) # 100000
````

### 4.2 Operadores relacionais

Comparam valores e produzem um resultado booleano (`True` ou `False`): `==` igual; `!=` diferente; `>` maior; `<` menor; `>=` maior ou igual; `<=` menor ou igual.

````python
idade = 18

print(idade >= 18)    # True
print(idade == 18)    # True
print(idade != 20)    # True
print(idade > 18)     # False
print(idade < 18)     # False
print(idade <= 17)    # False
````

### 4.3 Operadores lógicos

`and` (E) exige que as condições sejam verdadeiras; `or` (OU) exige pelo menos uma condição verdadeira; `not` (NÃO) inverte o valor lógico. Com booleanos, `^` (OU Exclusivo, XOR) retorna `True` quando exatamente um valor é verdadeiro.

````python
idade = 20
tem_carteira = True

print(idade >= 18 and tem_carteira)     # True
print(idade < 18 or tem_carteira)       # True
print(not tem_carteira)                 # False

valor1 = True
valor2 = False
print(valor1 ^ valor2)                  # True
print(valor1 ^ valor1)                  # False
````

## 5. IF, ELIF e ELSE

O que é? Estruturas condicionais permitem que o programa escolha quais comandos executar de acordo com condições. `if` testa a primeira condição, `elif` testa condições adicionais e `else` executa quando nenhuma condição anterior foi satisfeita.

**Estrutura do comando**

```python
if condição:
    comando
elif outra_condição:
    comando
else:
    comando
```

**Exemplo prático**

````python
nota = float(input("Digite sua nota: "))

if nota >= 7:
    print("Aprovado")
elif nota >= 5:
    print("Recuperação")
else:
    print("Reprovado")
````

Entenda o exemplo: Se a nota `for` 7 ou mais, o resultado será aprovado. Se `for` pelo menos 5 e menor que 7, será recuperação. Abaixo de 5, será reprovado. A ordem das condições importa.

## 6. WHILE e FOR

São estruturas de repetição (laços), usadas para executar um bloco de comandos mais de uma vez.

### 6.1 WHILE

O `while` repete os comandos enquanto a condição `for` verdadeira.

```python
while condição:
    comando
```

````python
contador = 1

while contador <= 5:
    print(contador)
    contador += 1
````

O contador começa em 1 e aumenta uma unidade a cada repetição. Quando chega a 6, a condição se torna falsa e o laço termina. Se a variável de controle não `for` atualizada corretamente, pode ocorrer um laço infinito.

### 6.2 FOR

O `for` percorre elementos de uma sequência ou repete comandos com base em uma sequência de valores.

```python
for variavel in sequência:
    comando
```

````python
for numero in range(1, 6):
    print(numero)
````

`range(1, 6)` produz os valores 1, 2, 3, 4 e 5. O limite final não é incluído.

Diferença prática: use `while` quando a repetição depende de uma condição; use `for` quando quiser percorrer uma sequência ou repetir com uma quantidade definida.

## 7. Funções

O que é? Funções são blocos reutilizáveis de código que realizam tarefas específicas. Elas ajudam a organizar o programa e evitar repetição.

**Estrutura do comando**

```python
def nome_da_funcao():
    comandos

nome_da_funcao()
```

**Exemplo prático**

````python
def saudacao():
    print("Olá! Seja bem-vindo!")

saudacao()
````

Entenda o exemplo: A palavra `def` define a função. A linha `saudacao()` chama a função, fazendo seu bloco ser executado.

**Funções com parâmetros e retorno**

```python
def nome_da_funcao(parametro):
    return resultado
```

````python
def somar(a, b):
    resultado = a + b
    return resultado

total = somar(10, 5)
print(total) # 15
````

Os parâmetros a e b recebem os valores enviados na chamada. `return` devolve o resultado para quem chamou a função.

## 8. Listas

O que é? Listas armazenam vários elementos em uma única estrutura. São mutáveis, isto é, podem ser alteradas depois de criadas, e são representadas por colchetes.

**Estrutura do comando**

```python
nome_da_lista = [elemento1, elemento2, elemento3]
```

**Exemplo prático**

````python
frutas = ["Maçã", "Banana", "Uva"]

print(frutas)
print(frutas[0])   # Maçã
print(frutas[1])   # Banana
````

Entenda o exemplo: Os índices começam em zero: o índice 0 acessa o primeiro elemento, o índice 1 acessa o segundo e assim por diante.

**Operações comuns**

````python
frutas.append("Laranja")    # Adiciona ao final
frutas.remove("Banana")     # Remove pelo valor
frutas.pop()                # Remove o último elemento
print(len(frutas))          # Quantidade de elementos
````

## 9. Tuplas

O que é? Tuplas armazenam uma sequência de elementos e, ao contrário das listas, são imutáveis: não permitem alterar, adicionar ou remover elementos diretamente após a criação. São normalmente representadas por parênteses.

**Estrutura do comando**

```python
nome_da_tupla = (elemento1, elemento2, elemento3)
```

**Exemplo prático**

````python
dias = ("Segunda", "Terça", "Quarta", "Quinta")

print(dias)
print(dias[0])
print(dias[2])
````

Entenda o exemplo: Assim como nas listas, os índices começam em zero. Uma tupla de um único elemento precisa de vírgula: `(5,)`.

## 10. Dicionários

O que é? Dicionários armazenam informações em pares de chave e valor. Cada chave identifica o valor associado. São representados por chaves.

**Estrutura do comando**

```python
nome_do_dicionario = {
    "chave1": valor1,
    "chave2": valor2
}
```

**Exemplo prático**

````python
aluno = {
    "nome": "Rafael",
    "idade": 20,
    "curso": "Matemática Aplicada e Computacional"
}

print(aluno["nome"])
print(aluno["idade"])
print(aluno["curso"])
````

Entenda o exemplo: As chaves são nome, idade e curso. Os valores associados são Rafael, 20 e Matemática Aplicada e Computacional.

**Operações comuns**

````python
aluno["nota"] = 9.0         # Adiciona uma chave
aluno["idade"] = 21         # Atualiza um valor
aluno.pop("curso")          # Remove a chave e seu valor
print(aluno)
````

## 11. Vetores

O que é? Em fundamentos de programação, um vetor é uma estrutura unidimensional, organizada em posições sequenciais. Em Python introdutório, uma lista pode representar esse tipo de estrutura.

**Estrutura do comando**

```python
vetor = [elemento1, elemento2, elemento3]
```

**Exemplo prático**

````python
numeros = [10, 20, 30, 40, 50]

for i in range(len(numeros)):
    print(numeros[i])
````

Entenda o exemplo: `len(numeros)` informa a quantidade de elementos. `range()` gera os índices e `numeros[i]` acessa o elemento na posição indicada.

**Acessando posições específicas**

````python
numeros = [10, 20, 30, 40, 50]

print(numeros[0])   # 10
print(numeros[2])   # 30
print(numeros[4])   # 50
````

## 12. Matrizes

O que é? Matrizes organizam dados em linhas e colunas. Uma matriz bidimensional pode ser representada em Python por uma lista cujos elementos também são listas.

**Estrutura do comando**

```python
matriz = [
    [elemento1, elemento2],
    [elemento3, elemento4]
]
```

**Exemplo prático**

````python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matriz[0][0])    # 1
print(matriz[1][2])    # 6
print(matriz[2][1])    # 8
````

Entenda o exemplo: O primeiro índice indica a linha e o segundo indica a coluna. Como os índices começam em zero, `matriz[1][2]` acessa a segunda linha e a terceira coluna.

**Percorrendo todos os elementos**

````python
for linha in matriz:
    for elemento in linha:
        print(elemento)
````

O primeiro `for` percorre cada linha. O segundo percorre os elementos da linha atual.

````python
def conferir(nome, seu, esperado, tol=0.01):
    """Compara a sua resposta com a de referência (tolerância `tol`). Não revela a resposta certa."""
    if seu is None or (np.ndim(seu) > 0 and any(v is None for v in np.ravel(seu))):
        print(f'[pendente] {nome}: ainda não preenchido'); return
    try:
        ok = np.allclose(np.asarray(seu, dtype=float), np.asarray(esperado, dtype=float), atol=tol, rtol=0)
    except (TypeError, ValueError):
        ok = False
    print(('[ok] ' if ok else '[revise] ') + nome + ('' if ok else ': confira o cálculo e rode de novo'))

def conferir_igual(nome, seu, esperado):
    """Versão para textos e dicionários."""
    if seu is None or (isinstance(seu, dict) and any(v is None for v in seu.values())):
        print(f'[pendente] {nome}: ainda não preenchido'); return
    if seu == esperado:
        print('[ok] ' + nome)
    else:
        errados = [k for k in esperado if seu.get(k) != esperado[k]] if isinstance(esperado, dict) else ''
        print('[revise] ' + nome + (f': revise {errados}' if errados else ': confira a resposta'))
````

Os comandos permitem reconhecer o formato (18 linhas e 5 colunas), os nomes, tipos e quantidade de valores presentes. `describe()` calcula medidas descritivas; o ID não entra nesse resumo porque é apenas um identificador.

````python
print("Primeiras linhas:\n", df.head())
print("Ultimas linhas:\n", df.tail())
print("Linhas e colunas:", df.shape)
print("Colunas:", df.columns)
df.info()   # Tipos e quantidade de valores presentes
print(df[["horas_estudo", "frequencia", "nota"]].describe())
````

## 2.3. Seleção de linhas e colunas

Uma coluna entre colchetes simples retorna uma Series; uma lista de colunas retorna um DataFrame. `loc` seleciona por rótulos e inclui o último rótulo de uma fatia. `iloc` seleciona por posições, excluindo a posição final. Os dois últimos comandos retornam as mesmas linhas e colunas.

````python
print(df["nota"])                      # Uma Series
print(df[["curso", "nota"]])           # Um DataFrame
print(df.loc[0:2, ["id", "nota"]])     # Rotulos
print(df.iloc[0:3, [0, 4]])            # Posicoes
````

## 2.4. Filtros e ordenação

O filtro `alto` seleciona notas de pelo menos 7 e frequência de pelo menos 80%. `atencao` seleciona notas abaixo de 7 ou ausentes, identificadas por `isna()`. Usa-se `&`, `|` e `~` no lugar de `and`, `or` e `not` para combinar condições linha a linha. `ascending=False` ordena da maior para a menor nota.

````python
print(df[df["curso"] == "Matematica"])
alto = (df["nota"] >= 7) & (df["frequencia"] >= 0.8)
print(df[alto])
atencao = (df["nota"] < 7) | (df["nota"].isna())
print(df[atencao])
print(df.sort_values("nota", ascending=False).head())
````

## 3. Qualidade dos dados

O pandas representa uma nota numérica ausente por `NaN`. `duplicated()` e `isna()` geram valores booleanos; `sum()` conta os verdadeiros como 1. O filtro `fora` procura notas registradas fora de 0 a 10. Uma tabela vazia indica que nenhuma nota atende ao filtro. Nesta base, as contagens de duplicatas e IDs repetidos são zero, e há duas notas ausentes.

````python
print("Duplicatas:", df.duplicated().sum())
print("IDs repetidos:", df["id"].duplicated().sum())
print("Ausentes:\n", df.isna().sum())
fora = df["nota"].notna() & ~df["nota"].between(0, 10)
print("Notas fora de 0 a 10:\n", df[fora])
````

## 3.1. Tratamento de valores ausentes

Uma nota ausente não é uma nota zero. `df` mantém os 18 registros; `df_com_notas` é uma cópia com as 16 notas disponíveis. O argumento `subset=['nota']` limita a exclusão às notas ausentes. `fillna()` permite substituir ausências quando existe uma justificativa; neste problema, não substituímos as notas.

````python
pendentes = df[df["nota"].isna()]
print(pendentes[["id", "curso"]])

df_com_notas = df.dropna(subset=["nota"]).copy()
print("Linhas e colunas:", df_com_notas.shape)
````

## 3.2. Criação de colunas

A frequência original está entre 0 e 1; a multiplicação por 100 a expressa em percentual. `apply()` aplica a função a cada nota e classifica os valores com o limite didático de 7, sem representar uma regra oficial de aprovação. As novas colunas serão usadas no agrupamento e no gráfico de dispersão.

````python
df_com_notas["frequencia_pct"] = df_com_notas["frequencia"] * 100
def classificar(nota):
    if nota >= 7:
        return "Nota >= 7"
    return "Nota < 7"
df_com_notas["situacao"] = df_com_notas["nota"].apply(classificar)
print(df_com_notas[["frequencia_pct", "nota", "situacao"]].head())
````

## 4. Agrupamento dos dados

`groupby('curso')` separa os registros por curso. Primeiro calculamos uma mediana para cada grupo. Com `agg()`, calculamos várias medidas de uma vez; `std` usa n−1 na variância por padrão. `value_counts()` conta as notas de cada categoria criada anteriormente. `round(2)` arredonda os valores exibidos.

````python
print(df_com_notas.groupby("curso")["nota"].median())
resumo = df_com_notas.groupby("curso")["nota"].agg(
    ["count", "mean", "median", "std", "min", "max"]
)
print(resumo.round(2))
print(df_com_notas.groupby("curso")["situacao"].value_counts())
````

## 4.1. Frequência relativa de notas ausentes

Ao sortear um registro de um curso nesta base, a probabilidade de a nota estar ausente é dada pela razão entre as notas ausentes e os registros desse curso. Em Matemática, 1/6 ≈ 16,67%. `size` conta todas as linhas; `count` conta apenas valores disponíveis. Multiplica-se a razão por 100 para expressá-la em percentual.

````python
cobertura = df.groupby("curso")["nota"].agg(["size", "count"])
cobertura["sem_nota"] = cobertura["size"] - cobertura["count"]
cobertura["pct_sem_nota"] = 100 * cobertura["sem_nota"] / cobertura["size"]
print(cobertura.round(2))
````

## 5. Matplotlib / Gráfico de barras

O eixo x contém os cursos (índice de `cobertura`), e o eixo y contém os percentuais de notas ausentes. `figsize` define largura e altura em polegadas, `color` define a cor e `bar_label` mostra os valores nas barras. `title`, `xlabel` e `ylabel` rotulam o gráfico, `ylim` limita o eixo y, `tight_layout()` ajusta o espaço e `show()` exibe a figura.

````python
plt.rcParams["font.size"] = 13   # Tamanho da fonte
plt.figure(figsize=(6, 4))       # Largura e altura
barras = plt.bar(cobertura.index, cobertura["pct_sem_nota"],
                 color="#358FD0")
plt.bar_label(barras, fmt="%.1f%%") # Valores nas barras
plt.title("Notas ausentes por curso")
plt.xlabel("Curso")
plt.ylabel("Notas ausentes (%)")
plt.ylim(0, 100)
plt.tight_layout()               # Ajusta o espaco
plt.show()                       # Exibe o grafico
````

### 5.1. Comparação dos cursos

Matemática e Estatística possuem uma nota ausente em seis registros (16,7%). Computação tem todas as notas disponíveis.

## 5.2. Histograma

O histograma resume a distribuição das notas em intervalos. O parâmetro `bins=5` divide o intervalo entre a menor e a maior nota em cinco classes de mesma largura.

````python
plt.figure(figsize=(6, 4))
plt.hist(df_com_notas["nota"], bins=5, color="#358FD0", edgecolor="white")
plt.title("Distribuição das notas")
plt.xlabel("Nota")
plt.ylabel("Frequência absoluta")
plt.tight_layout()
plt.show()
````

### 5.3. Distribuição das notas

As notas disponíveis variam de 5,5 a 9,6. Os quatro primeiros intervalos contêm três notas cada, e o último contém quatro.

## 5.4. Gráfico de dispersão

Cada ponto representa um estudante, com frequência em percentual no eixo x e nota no eixo y. O laço percorre os cursos; `label` define o nome de cada grupo e `legend()` exibe a legenda.

````python
plt.figure(figsize=(6, 4))
for curso, grupo in df_com_notas.groupby("curso"):
    plt.scatter(grupo["frequencia_pct"], grupo["nota"], label=curso)
plt.title("Frequência e nota")
plt.xlabel("Frequência (%)")
plt.ylabel("Nota")
plt.legend()
plt.tight_layout()
plt.show()
````

### 5.5. Relação entre frequência e nota

Nesta base fictícia, as notas maiores tendem a acompanhar frequências mais altas. Essa associação, por si só, não demonstra uma relação de causa e efeito.

### 6. Interpretação e proposta de ação

| Curso | Notas disponíveis | Mediana | Desvio padrão | Ausentes |
|---|---:|---:|---:|---:|
| Computação | 6 | 7,85 | 1,29 | 0,0% |
| Estatística | 5 | 7,80 | 1,38 | 16,7% |

As medianas e os desvios padrão das notas disponíveis são próximos entre os cursos.

**Avaliação:** acompanhar o percentual de notas ausentes por curso, com meta de 0%, e repetir a análise após o registro.

As diferenças entre as medianas descrevem somente os registros disponíveis nesta base pequena e fictícia; não medem a qualidade dos cursos.

## 6.1. Junção de tabelas

O resultado final combina as medidas das notas com o percentual de notas ausentes. `reset_index()` transforma o índice de cursos em uma coluna. `merge()` junta as tabelas pela coluna indicada em `on='curso'`, mantendo as chaves presentes em ambas por padrão.

````python
ausencias = cobertura[["pct_sem_nota"]].reset_index()
resultado = pd.merge(resumo.reset_index(), ausencias, on="curso")
print(resultado.round(2))
````

## 6.2. Exportação dos resultados

O arquivo `resumo_por_curso.csv` reúne as medidas de posição e dispersão e a proporção de notas ausentes. `to_csv()` exporta a tabela e `index=False` evita gravar o índice numérico como coluna extra.

````python
resultado.round(2).to_csv(
    "resumo_por_curso.csv",
    index=False
)
print("Arquivo resumo_por_curso.csv gerado.")
````
