### Exercício 3: Biblioteca de funções

*Tópicos: def, return, if/elif/else, while, for, range, resto da divisão e dicionários.*

Implemente primeiro três funções. Na classificação, considere o critério didático: nota ≥ 7, aprovado; 5 ≤ nota < 7, recuperação; nota < 5, reprovado. O fatorial recebe um inteiro não negativo e a soma de pares recebe um limite inteiro não negativo.

```python
def classificar(nota):
    if nota >= 7:
        return "Aprovado"
    elif nota >= 5:
        return "Recuperação"
    else:
        return "Reprovado"


def fatorial(n):
    resultado = 1
    while n > 1:
        resultado *= n
        n -= 1
    return resultado


def soma_pares(limite):
    total = 0
    for i in range(1, limite + 1):
        if i % 2 == 0:
            total += i
    return total
```

```python
print(classificar(6.5))                      # Recuperação
print(fatorial(5))                           # 120
print(soma_pares(10))                        # 30
```

**Complementar:** implemente `contar_situacoes(notas)`. Use a função `classificar` para retornar um dicionário com a quantidade de notas em cada situação.

```python
def contar_situacoes(notas):
    contagem = {"Aprovado": 0, "Recuperação": 0, "Reprovado": 0}
    for nota in notas:
        situacao = classificar(nota)
        contagem[situacao] += 1
    return contagem
```

```python
print(contar_situacoes([8, 5, 3, 7, 9, 4]))  # {'Aprovado': 3, 'Recuperação': 1, 'Reprovado': 2}
```

### Exercício 5: Relatório de vendas da cantina

Uma cantina fictícia registra suas vendas ao longo de três dias. Cada venda é uma tupla `(produto, categoria, preço, quantidade)`; a quantidade é o total vendido no período. A matriz registra as unidades vendidas por dia, com os produtos na mesma ordem da lista.

1. Implemente `receita_por_categoria(vendas)`: calcule a receita total (preço × quantidade) de cada categoria em um dicionário.
2. Implemente `produto_mais_vendido(vendas)`: retorne o nome do produto com maior quantidade total vendida.
3. Implemente `melhor_dia(matriz)`: retorne o número do dia (a partir de 1) com mais unidades vendidas.
4. Informe se cada categoria atingiu a meta de receita de R$ 200 (valor maior ou igual a 200).

Considere os dados fornecidos; em caso de empate, mantenha o primeiro item encontrado.

```python
vendas_cantina = [
    ("Café", "bebida", 5.0, 30),
    ("Suco", "bebida", 7.0, 12),
    ("Pão de queijo", "salgado", 6.0, 25),
    ("Coxinha", "salgado", 8.0, 18),
    ("Bolo", "doce", 9.0, 10)
]

# cada linha = um produto (mesma ordem da lista acima); cada coluna = um dia
unidades_dia = [
    [10, 12, 8],
    [4, 3, 5],
    [8, 9, 8],
    [6, 5, 7],
    [3, 4, 3]
]

META = 200
```

```python
def receita_por_categoria(vendas):
    receitas = {"bebida": 0, "salgado": 0, "doce": 0}
    for venda in vendas:
        categoria = venda[1]
        receitas[categoria] += venda[2] * venda[3]
    return receitas


def produto_mais_vendido(vendas):
    melhor = vendas[0]
    for venda in vendas:
        if venda[3] > melhor[3]:
            melhor = venda
    return melhor[0]


def melhor_dia(matriz):
    melhor = 0
    maior_total = 0
    for j in range(len(matriz[0])):
        total = 0
        for i in range(len(matriz)):
            total += matriz[i][j]
        if total > maior_total:
            maior_total = total
            melhor = j
    return melhor + 1


# Relatório (tarefa 4)
receitas = receita_por_categoria(vendas_cantina)
categorias = ["bebida", "salgado", "doce"]

for categoria in categorias:
    if receitas[categoria] >= META:
        print(categoria, "bateu a meta")
    else:
        print(categoria, "ficou abaixo da meta")
```

```python
print(receita_por_categoria(vendas_cantina))   # {'bebida': 234.0, 'salgado': 294.0, 'doce': 90.0}
print(produto_mais_vendido(vendas_cantina))    # Café
print(melhor_dia(unidades_dia))                # 2

# Relatório esperado (tarefa 4):
# bebida bateu a meta
# salgado bateu a meta
# doce ficou abaixo da meta
```

### Exercício 4: Estatística "na mão"

Implemente média, variância amostral e desvio padrão utilizando laços e operações aritméticas. Use as funções prontas do NumPy apenas para conferir a solução. Considere listas com pelo menos dois valores para a variância.

```python
def media(valores):
    soma = 0
    for v in valores:
        soma += v
    return soma / len(valores)


def variancia(valores):
    m = media(valores)
    soma = 0
    for v in valores:
        soma += (v - m) ** 2
    return soma / (len(valores) - 1)


def desvio_padrao(valores):
    return variancia(valores) ** 0.5
```

```python
x = [2, 3, 4, 4, 5, 6, 8]
print(media(x))          # 4.5714
print(variancia(x))      # 3.9524
print(desvio_padrao(x))  # 1.9881
xn = np.array(x)
print(xn.mean(), xn.var(ddof=1), xn.std(ddof=1))
```

**Complementar (desafio):** implemente a correlação de Pearson com laços e compare o resultado com `np.corrcoef(a, b)[0, 1]`. Considere listas de mesmo tamanho e com dispersão não nula.

```python
def correlacao(a, b):
    media_a = media(a)
    media_b = media(b)
    num = 0
    soma_a = 0
    soma_b = 0
    for i in range(len(a)):
        da = a[i] - media_a
        db = b[i] - media_b
        num += da * db
        soma_a += da ** 2
        soma_b += db ** 2
    return num / (soma_a * soma_b) ** 0.5
```

```python
a = [1, 2, 3, 4, 5]
b = [2, 1, 4, 3, 5]
print(correlacao(a, b))  # 0.8
print(round(np.corrcoef(a, b)[0, 1], 2))
```

**Complementar:** implemente a identificação dos valores atípicos pelas cercas do IQR, considerando os quartis fornecidos.

```python
def atipicos(valores, q1, q3):
    iqr = q3 - q1
    inferior = q1 - 1.5 * iqr
    superior = q3 + 1.5 * iqr
    fora = []
    for v in valores:
        if v < inferior:
            fora.append(v)
        if v > superior:
            fora.append(v)
    return fora
```

```python
print(atipicos(x, 3.5, 5.5))  # []
print(atipicos([2, 3, 4, 4, 5, 6, 24], 3.5, 5.5))  # [24]
```

**Complementar A — Binomial:** calcule a proporção de estudantes com sono menor que 6 h. Suponha 20 estudantes independentes, com essa mesma probabilidade de sono curto. Calcule $E[X]=np$ e $P(X\geq8)$ usando $1-P(X\leq7)$. A função `prob_binomial(n, k, p)` fornece $P(X=k)$; `comb(n, k)` calcula o coeficiente binomial. Some as probabilidades de 0 a 7 para obter $P(X\leq7)$. Resultados esperados: 5 e 0,1018.

```python
from math import comb

# Funcao auxiliar fornecida: P(X = k)
def prob_binomial(n, k, p):
    return comb(n, k) * p ** k * (1 - p) ** (n - k)

p_sono = (np.array(sono) < 6).mean()
esperado = 20 * p_sono
acumulada = 0
for k in range(8):
    acumulada += prob_binomial(20, k, p_sono)
p_binomial = 1 - acumulada
print("p_sono:", p_sono)
print("Esperado:", esperado)
print("P(X >= 8):", round(p_binomial, 4))
```

**Complementar B — Poisson:** em outra situação simulada, um setor de apoio acadêmico recebe em média 2,3 solicitações por dia. Suponha uma contagem diária modelada por Poisson, com $\lambda=2{,}3$. Calcule $P(X=0)$ e $P(X\geq5)=1-P(X\leq4)$. A função `prob_poisson(k, lam)` fornece $P(X=k)$; `factorial(k)` calcula $k!$. Resultados: 0,1003 e 0,0838.

```python
from math import factorial

# Funcao auxiliar fornecida: P(X = k)
def prob_poisson(k, lam):
    return np.exp(-lam) * lam ** k / factorial(k)

lam = 2.3
p_zero = prob_poisson(0, lam)
acumulada = 0
for k in range(5):
    acumulada += prob_poisson(k, lam)
p_poisson = 1 - acumulada
print("P(X = 0):", round(p_zero, 4))
print("P(X >= 5):", round(p_poisson, 4))
```

**Complementar C — Normal:** suponha um modelo normal hipotético de horas de sono, com média 7 h e desvio padrão 1 h. Calcule a probabilidade de menos de 6 h de sono usando a função fornecida `normal_cdf(x, media, desvio)`, que retorna a probabilidade acumulada até `x`. Compare com a proporção empírica, que pode ser calculada a partir da lista `sono`. Resultado do modelo: 0,1587.

```python
from math import erf

# Funcao auxiliar fornecida: P(X <= x)
def normal_cdf(x, media, desvio):
    z = (x - media) / (desvio * np.sqrt(2))
    return (1 + erf(z)) / 2

p_normal = normal_cdf(6, 7, 1)
print("Normal: P(X < 6) =", round(p_normal, 4))
print("Proporção observada:", (np.array(sono) < 6).mean())
```

Os modelos dependem de suas hipóteses. A proporção observada de sono curto é 0,25, diferente de 0,1587 no modelo normal hipotético.

### Exercício 3: Preparação dos dados

*Tópicos: funções, isna, dropna, copy, apply e Matplotlib.*

1. Utilize `isna().sum()` para conferir quantos valores estão ausentes em cada coluna da base original.
2. Selecione as colunas `survived, pclass, sex, age, fare` em uma cópia chamada `df_dados`. Crie outra cópia, `df_com_idade`, apenas com as idades disponíveis, utilizando `dropna(subset=["age"])`.
3. Implemente somente `faixa_etaria(idade)`: abaixo de 13, criança; de 13 a menos de 30, jovem; de 30 a menos de 60, adulto; a partir de 60, idoso. Os limites são didáticos. Crie a coluna `faixa` de `df_com_idade` usando `apply()`.

A idade ausente não equivale a zero. Preserve todos os 891 registros para analisar classe e sexo e utilize os 714 registros com idade conhecida apenas nas análises que necessitam dessa informação.

```python
print(df_bruto.isna().sum())

df_dados = df_bruto[["survived", "pclass", "sex", "age", "fare"]].copy()
df_com_idade = df_dados.dropna(subset=["age"]).copy()
```

**3.** Implemente a função a seguir e aplique-a à coluna de idades.

```python
def faixa_etaria(idade):
    if idade < 13:
        return "criança"
    elif idade < 30:
        return "jovem"
    elif idade < 60:
        return "adulto"
    else:
        return "idoso"

df_com_idade["faixa"] = df_com_idade["age"].apply(faixa_etaria)
```

```python
# Confira as duas tabelas depois de completar a parte principal.
print(df_dados.shape)       # (891, 5)
print(df_com_idade.shape)   # (714, 6)
print(df_com_idade["age"].isna().sum())  # 0
print(df_com_idade["faixa"].value_counts())
```

**Complementar:** construa um gráfico de barras das quantidades de passageiros por faixa etária.

### Exercício 4: Comparação das taxas de sobrevivência

*Tópicos: groupby, agg, probabilidade condicional e gráfico de barras.*

Execute a preparação deste bloco. Esta questão pode ser resolvida independentemente do exercício 3, pois a análise por sexo e classe utiliza a base original `df_bruto`. Investigue como a proporção de sobreviventes varia entre os grupos registrados.

1. Implemente `taxa_por_grupo(df, coluna)`. Agrupe pela coluna recebida e use `agg(["count", "sum"])` na coluna `survived`. A contagem corresponde aos passageiros e a soma, aos sobreviventes. Crie a coluna `taxa_pct`, com o percentual de sobreviventes em cada grupo.
2. Utilize a função para comparar sexo e classe, mantendo os 891 registros, inclusive aqueles sem idade.
3. Construa um gráfico de barras dos percentuais por classe, com títulos e rótulos.
4. Interprete os resultados. Qual classe apresenta maior percentual de sobrevivência? Qual apresenta menor? Explique a relação do cálculo com a probabilidade condicional no conjunto observado.

```python
def taxa_por_grupo(df, coluna):
    resumo = df.groupby(coluna)["survived"].agg(["count", "sum"])
    resumo["taxa_pct"] = 100 * resumo["sum"] / resumo["count"]
    return resumo

taxas_sexo = taxa_por_grupo(df_bruto, "sex")
taxas_classe = taxa_por_grupo(df_bruto, "pclass")
print(taxas_sexo.round(2))
print(taxas_classe.round(2))

plt.figure(figsize=(6, 4))
plt.bar(taxas_classe.index, taxas_classe["taxa_pct"])
plt.xticks([1, 2, 3])
plt.xlabel("Classe")
plt.ylabel("Sobrevivência (%)")
plt.title("Sobrevivência por classe")
plt.ylim(0, 100)
plt.show()
```

**Interpretação:** a primeira classe tem o maior percentual (62,96%); a terceira, o menor (24,24%). A primeira possui 136 sobreviventes entre 216 passageiros. Assim, ao sortear um registro dessa classe, $P(\text{sobreviveu}\mid\text{classe 1})=136/216\approx0{,}6296$. As diferenças são descritivas e não demonstram causalidade.

1. Apresente a quantidade de passageiros, a proporção geral de sobreviventes e a quantidade de idades ausentes.
2. Compare as taxas de sobrevivência por faixa etária, utilizando apenas o conjunto com idade disponível. Informe o denominador de cada grupo.
3. Agrupe por classe as idades disponíveis e calcule mediana e desvio padrão amostral. Junte essa tabela à tabela de sobrevivência por classe com `reset_index()` e `pd.merge()`.
4. Construa um histograma de idades disponíveis e um gráfico de dispersão entre idade e valor da passagem, com cores por classe e legenda.
5. Exporte a tabela final como `resumo_titanic.csv` usando `to_csv(index=False)`.
6. Interprete os resultados por classe e faixa etária. Explique como as idades ausentes limitam a análise e indique qual informação adicional seria útil para investigar as diferenças observadas. Não atribua as diferenças a uma causa apenas com estes dados.

Consulte as funções do Guia de Apoio e os exemplos dos slides. A questão é uma análise descritiva e permite diferentes soluções corretas.

```python
print("Total de passageiros:", len(df_dados))
print("Sobreviventes (%):", round(100 * df_dados["survived"].mean(), 2))
print("Idades ausentes:", df_dados["age"].isna().sum())

taxas_faixa = taxa_por_grupo(df_com_idade, "faixa")
print(taxas_faixa.round(2))

taxas_classe = taxa_por_grupo(df_dados, "pclass").reset_index()
idades_classe = df_com_idade.groupby("pclass")["age"].agg(["median", "std"]).reset_index()
resultado = pd.merge(taxas_classe, idades_classe, on="pclass")
print(resultado.round(2))

plt.figure(figsize=(6, 4))
plt.hist(df_com_idade["age"], bins=20, edgecolor="white")
plt.xlabel("Idade (anos)")
plt.ylabel("Passageiros")
plt.title("Distribuição das idades disponíveis")
plt.show()

plt.figure(figsize=(7, 4))
for classe, grupo in df_com_idade.groupby("pclass"):
    plt.scatter(grupo["age"], grupo["fare"], label=f"Classe {classe}", alpha=0.6)
plt.xlabel("Idade (anos)")
plt.ylabel("Valor da passagem")
plt.title("Idade e valor da passagem por classe")
plt.legend()
plt.tight_layout()
plt.show()

resultado.round(2).to_csv("resumo_titanic.csv", index=False)
print("Arquivo resumo_titanic.csv gerado.")
```

**Interpretação:** entre todos os registros, a primeira classe possui 62,96% de sobreviventes, contra 24,24% na terceira. As medianas das idades disponíveis são 37, 29 e 24 anos nas classes 1, 2 e 3, respectivamente. Entre os 714 passageiros com idade registrada, crianças apresentam 40 sobreviventes em 69 registros (57,97%); idosos, 7 em 26 (26,92%). As 177 idades ausentes podem alterar as comparações por faixa etária. Para investigar as diferenças entre os grupos, seria útil conhecer informações individuais sobre o acesso a botes e as circunstâncias do embarque. Os dados não demonstram, por si só, causalidade.
