# 4. Estatística à mão: 7 alunos

> 🎯 **Ao final:** calcular média, mediana, quartis, variância e desvio padrão com papel e lápis, e explicar o `n − 1`.

Com 15 mil linhas ninguém confere uma fórmula. Com 7, todo mundo confere.

Horas diárias em redes sociais de 7 alunos: $x = [2, 3, 4, 4, 5, 6, 8]$.

````python
x = np.array([2, 3, 4, 4, 5, 6, 8])
n = len(x)
````

## 4.1 Média, mediana e moda

$$\bar{x} = \frac{\sum x_i}{n} \qquad \text{mediana} = \text{valor central dos dados ordenados} \qquad \text{moda} = \text{valor mais frequente}$$

````python
media = x.sum() / n
mediana = np.sort(x)[n // 2]
moda = pd.Series(x).mode()[0]
print(f'média = {x.sum()}/{n} = {media:.2f}  (numpy: {x.mean():.2f})')
print(f'mediana = {mediana}  (numpy: {np.median(x)})')
print(f'moda = {moda}')
````

### 🧠 A média é sensível a outliers? Vamos **ver**.
Troque o aluno que usa 8h por um que usa 24h. Previsão: o que acontece com a média? E com a mediana?

````python
y = x.copy(); y[-1] = 24
pd.DataFrame({'original': [x.mean(), np.median(x)], 'com outlier (24h)': [y.mean(), np.median(y)]},
             index=['média', 'mediana']).round(2)
````

Um único valor mudou a média em mais de 2 horas e não mexeu na mediana. Isso vale para amostras pequenas; com 15 mil linhas, um valor só pesa pouco, mas **muitos** valores extremos pesam do mesmo jeito.

## 4.2 Percentis, quartis e IQR

O **percentil p** é o valor abaixo do qual ficam aproximadamente p% dos dados. Os **quartis** são os percentis 25, 50 e 75 ($Q_1$, $Q_2$ = mediana, $Q_3$).

Método usado por `numpy`/`pandas`: posição (a partir de 0) $= p\,(n-1)$, com interpolação linear entre vizinhos.

$$\text{IQR} = Q_3 - Q_1 \quad\text{(amplitude dos 50\% centrais)}$$

````python
def quantil(v, p):
    v = np.sort(v); pos = p * (len(v) - 1); i = int(pos); f = pos - i
    return v[i] + f * (v[min(i + 1, len(v) - 1)] - v[i])

q1, q3 = quantil(x, .25), quantil(x, .75)
print(f'Q1 = {q1}, Q3 = {q3}, IQR = {q3 - q1}')
print('numpy:', np.percentile(x, [25, 75]))
print('IQR com outlier:', np.percentile(y, 75) - np.percentile(y, 25), '(original:', q3 - q1, ')')
````

## 4.3 Variância e desvio padrão

Ideia: medir o quanto os valores se afastam da média. Se somarmos os desvios $x_i - \bar{x}$, o resultado é **sempre zero**; por isso elevamos ao quadrado.

$$s^2 = \frac{\sum (x_i - \bar{x})^2}{n-1} \qquad s = \sqrt{s^2}\ \ (\text{mesma unidade dos dados})$$

````python
tab = pd.DataFrame({'x': x, 'x − média': x - media, '(x − média)²': (x - media) ** 2}).round(3)
print(tab.to_string(index=False))
print('\nSoma dos desvios:', round((x - media).sum(), 10))
ss = ((x - media) ** 2).sum()
print(f'Soma dos quadrados = {ss:.3f} | s² = {ss:.3f}/{n - 1} = {ss / (n - 1):.3f} | s = {np.sqrt(ss / (n - 1)):.3f}')
````

### Por que dividir por $n-1$ e não por $n$?

Usamos $\bar{x}$ (calculada **da própria amostra**) como centro. Ela é o ponto que torna a soma dos quadrados a menor possível para essa amostra, então os desvios saem *menores* que os desvios em relação à média verdadeira. Dividir por $n$ **subestima** a variância. Vamos medir esse efeito: sorteamos 20.000 amostras de 5 alunos.

````python
pop = df['sono'].to_numpy()
amostras = [rng.choice(pop, 5, replace=False) for _ in range(20000)]
print(f'Variância real (todos os alunos):   {pop.var(ddof=1):.3f}')
print(f'Média das variâncias com ÷ n:       {np.mean([a.var(ddof=0) for a in amostras]):.3f}')
print(f'Média das variâncias com ÷ (n−1):   {np.mean([a.var(ddof=1) for a in amostras]).round(3)}')
````

Com $n=5$, dividir por $n$ dá cerca de $\tfrac{4}{5} = 80\%$ do valor correto. Outra leitura: como os desvios somam zero, só $n-1$ deles são "livres" (**graus de liberdade**).

### Como se desenha um boxplot (à mão)

O boxplot resume os dados em **cinco números** e uma regra para marcar valores atípicos. Com os mesmos 7 alunos, $x=[2,3,4,4,5,6,8]$:

| Passo | Cálculo | Resultado |
|---|---|---|
| 1. Quartis | $Q_1$, mediana, $Q_3$ | 3,5 · 4 · 5,5 |
| 2. IQR | $Q_3 - Q_1$ | 2 |
| 3. Limites (cercas) | $Q_1 - 1{,}5\,\text{IQR}$ e $Q_3 + 1{,}5\,\text{IQR}$ | 0,5 e 8,5 |
| 4. Bigodes | menor e maior valor **dentro** dos limites | de 2 até 8 |
| 5. Outliers | valores **fora** dos limites, desenhados como pontos | nenhum |

Desenhe um eixo, a **caixa** de $Q_1$ a $Q_3$, um traço na mediana, os bigodes e os pontos isolados. Troque o 8 por 24 (como na Seção 4): o IQR continua 2 e o limite superior continua 8,5, então o 24 vira um **ponto isolado** e o bigode passa a terminar em 6, o maior valor dentro do limite. Confira o roteiro em código:

````python
def cinco_numeros(s):
    q1, med, q3 = s.quantile([.25, .5, .75]); i = q3 - q1
    lo, hi = q1 - 1.5 * i, q3 + 1.5 * i
    return pd.Series({'Q1': q1, 'mediana': med, 'Q3': q3, 'limite inf.': lo, 'limite sup.': hi,
                      'bigode inf.': s[s >= lo].min(), 'bigode sup.': s[s <= hi].max(),
                      'outliers': ((s < lo) | (s > hi)).sum()})

pd.DataFrame({'7 alunos': cinco_numeros(pd.Series(x)), 'com 24h': cinco_numeros(pd.Series(y)),
              'redes, High School': cinco_numeros(df.loc[df['nivel'] == 'High School', 'redes'])}).round(2)
````

# 6. Correlação e seus limites

> 🎯 **Ao final:** calcular e interpretar $r$, e dar três motivos para não confiar nele cegamente.

$$r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \, \sum (y_i - \bar{y})^2}} \qquad -1 \le r \le 1$$

O numerador é positivo quando $x$ e $y$ ficam **acima da média juntos** (ou abaixo juntos). Exemplo à mão: $x=[1,2,3,4,5]$, $y=[2,1,4,3,5]$.

````python
a = np.array([1, 2, 3, 4, 5]); b = np.array([2, 1, 4, 3, 5])
da, db = a - a.mean(), b - b.mean()
print('produtos:', da * db, '→ soma =', (da * db).sum())
print('r =', (da * db).sum() / np.sqrt((da ** 2).sum() * (db ** 2).sum()), '| numpy:', np.corrcoef(a, b)[0, 1].round(3))
````

### Três armadilhas

2. **Um ponto** pode criar ou destruir uma correlação.
3. Uma terceira variável (calor) pode causar as duas; sorvete não causa afogamento.

# 7. Probabilidade: o básico, Binomial e Poisson

> 🎯 **Ao final:** estimar probabilidades por frequência, calcular uma probabilidade condicional, e saber quando usar Binomial e quando usar Poisson.

## 7.1 O básico

- Probabilidade de um evento $A$: $0 \le P(A) \le 1$, e $P(\text{não } A) = 1 - P(A)$.
- Estimativa empírica: $P(A) \approx \dfrac{\text{vezes que } A \text{ ocorreu}}{\text{total}}$.
- **Condicional:** $P(A \mid B) = \dfrac{P(A \cap B)}{P(B)}$. Se $P(A \mid B) = P(A)$, os eventos são **independentes**.

````python
p_sono = (df['sono'] < 6).mean()
p_mental = (df['saude_mental'] < 60).mean()
p_cond = (df.loc[df['redes'] > 6, 'saude_mental'] < 60).mean()

print(f'P(dormir < 6h)                           = {p_sono:.3f}')
print(f'P(saúde mental < 60)                     = {p_mental:.3f}')
print(f'P(saúde mental < 60 | redes > 6h)        = {p_cond:.3f}')
````

Saber que o aluno usa mais de 6h de redes **dobra** a probabilidade de escore baixo: os eventos **não são independentes** (coerente com $r=-0{,}40$).

## 7.2 Distribuição Binomial

Contar **quantos sucessos em $n$ tentativas**, quando as tentativas são independentes e a probabilidade $p$ é a mesma em todas.

$$P(X=k) = \binom{n}{k} p^k (1-p)^{n-k} \qquad E[X]=np \qquad \mathrm{Var}[X]=np(1-p)$$

Exemplo: numa turma de $n=20$ alunos, $X$ = número de alunos com menos de 6h de sono ($p\approx 0{,}33$). À mão, $P(X=0) = (1-p)^{20}$.

### Como se desenha o gráfico de barras de uma distribuição

Eixo horizontal: os valores possíveis $k = 0, 1, 2, \dots$ Eixo vertical: a probabilidade. Para cada $k$, desenhe uma barra de altura $P(X=k)$, calculada pela fórmula. Para os **dados observados**, a altura é (quantas turmas tiveram exatamente $k$) ÷ (total de turmas). Se o modelo é bom, barras dos dados e pontos da fórmula ficam próximos. Na Poisson é igual, com $P(X=k)=e^{-\lambda}\lambda^k/k!$.

````python
n = 20
print('à mão  P(X=0) =', round((1 - p_sono) ** n, 5), '| scipy:', round(stats.binom.pmf(0, n, p_sono), 5))

contagens = (rng.choice(df['sono'].to_numpy(), (10000, n)) < 6).sum(axis=1)
k = np.arange(0, 17)
plt.bar(k - .2, np.bincount(contagens, minlength=17)[:17] / 10000, .4, label='10.000 turmas sorteadas dos dados')
plt.bar(k + .2, stats.binom.pmf(k, n, p_sono), .4, label=f'Binomial(n=20, p={p_sono:.2f})')
plt.xlabel('alunos com < 6h de sono na turma'); plt.ylabel('probabilidade'); plt.legend(); plt.show()

print(f'média simulada {contagens.mean():.2f} vs np = {n * p_sono:.2f}')
print(f'P(X ≥ 10) = {stats.binom.sf(9, n, p_sono):.4f}')
````

## 7.3 Distribuição de Poisson

Contar **quantas vezes um evento raro ocorre** em um intervalo (de tempo, espaço, grupo), a uma taxa média $\lambda$.

$$P(X=k) = \frac{e^{-\lambda}\lambda^k}{k!} \qquad E[X]=\mathrm{Var}[X]=\lambda$$

Esta base não tem contagens por intervalo de tempo, então vamos **construir** uma: embaralhamos os alunos, formamos grupos de 100 e contamos, em cada grupo, quantos dormem **menos de 4h** (evento raro, $p\approx 0{,}02$).

````python
emb = df.sample(frac=1, random_state=0)
raro = (emb['sono'] < 4).to_numpy()
raro = raro[:len(raro) // 100 * 100]
cont = raro.reshape(-1, 100).sum(axis=1)
lam = cont.mean()
print(f'{len(cont)} grupos | média = {lam:.2f} | variância = {cont.var(ddof=1):.2f}   ← Poisson: média ≈ variância')

k = np.arange(0, cont.max() + 1)
plt.bar(k, np.bincount(cont, minlength=len(k)) / len(cont), alpha=.6, label='observado nos dados')
plt.plot(k, stats.poisson.pmf(k, lam), 'o-', color='red', label=f'Poisson(λ={lam:.2f})')
plt.plot(k, stats.binom.pmf(k, 100, raro.mean()), 'x--', color='black', label='Binomial(100, p)')
plt.xlabel('alunos com < 4h de sono no grupo de 100'); plt.ylabel('proporção de grupos'); plt.legend(); plt.show()
````

A Poisson (vermelho) e a Binomial (preto) praticamente coincidem: **a Poisson é o limite da Binomial** quando $n$ é grande, $p$ é pequeno e $\lambda = np$.

| | Binomial | Poisson |
|---|---|---|
| Conta | sucessos em $n$ tentativas | ocorrências em um intervalo |
| Parâmetros | $n$ e $p$ | $\lambda$ (taxa média) |
| Limite superior | $n$ | nenhum |
| Média e variância | $np$ e $np(1-p)$ | ambas $\lambda$ |
| Exemplo real | alunos com sono curto numa turma de 20 | chamadas por hora numa central |
